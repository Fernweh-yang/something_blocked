```python
import os
import shutil
from sklearn.model_selection import train_test_split

# 定义数据路径
data_dir = 'CHASEDB1'
output_dir = 'data'

# 获取所有原图和mask路径
images = [f for f in os.listdir(data_dir) if f.endswith('.jpg')]
masks_1st = [f for f in os.listdir(data_dir) if f.endswith('_1stHO.png')]
masks_2nd = [f for f in os.listdir(data_dir) if f.endswith('_2ndHO.png')]

# 按8:2划分数据集
train_images, test_images = train_test_split(images, test_size=0.2, random_state=42)

# 创建输出文件夹结构
for split in ['train', 'test']:
    os.makedirs(os.path.join(output_dir, split, 'images'), exist_ok=True)
    os.makedirs(os.path.join(output_dir, split, 'mask1'), exist_ok=True)
    os.makedirs(os.path.join(output_dir, split, 'mask2'), exist_ok=True)

# 复制文件到相应文件夹
def copy_files(image_list, split):
    for image in image_list:
        # 获取图像文件名的基础名
        base_name = os.path.splitext(image)[0]
        # 复制原图
        shutil.copy(os.path.join(data_dir, image), os.path.join(output_dir, split, 'images', image))
        # 复制对应的mask
        mask_1st = base_name + '_1stHO.png'
        mask_2nd = base_name + '_2ndHO.png'
        shutil.copy(os.path.join(data_dir, mask_1st), os.path.join(output_dir, split, 'mask1', mask_1st))
        shutil.copy(os.path.join(data_dir, mask_2nd), os.path.join(output_dir, split, 'mask2', mask_2nd))

# 复制训练集和测试集
copy_files(train_images, 'train')
copy_files(test_images, 'test')

print("数据划分完成")
```
